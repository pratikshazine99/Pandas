```python
import pandas as pd

data = {
    "Name": ["Rahul", "Aman", "Priya", "Shad", "Neha", "Vikas"],
    "Age": [22, 24, 21, 25, 23, 28],
    "City": ["Jaipur", "Delhi", "Pune", "Jaipur", "Mumbai", "Delhi"],
    "Salary": [25000, 35000, 30000, 45000, 40000, 55000],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"]
}

df = pd.DataFrame(data)
```

```python
#Q.1 Arrange all employees by salary in descending order.
df.sort_values(by='Salary',ascending=False)
```

```python
#Q2 Find the employess receiving the highest salary.
df[df["Salary"]==df["Salary"].max()]
```

```python
#Q3 Find the  employess receiving the lowest salary. 
df[df["Salary"]==df["Salary"].min()]
```

```python
#Q4 Display only the first three employees using iloc
df.iloc[:3]
```

```python
#Q5. Create a Bonus column where salary above 40000 gets 15%, otherwise 5%.
df["Bonus"] = df["Salary"].apply(lambda x: x * 0.15 if x > 40000 else x * 0.05)
print(df)
```

```python
#Q6 Find the average salary of all employees.
df["Salary"].mean()
```

```python
#Q7 Rename Salary column to Monthly_Salary.
df.rename(columns={"Salary": "Monthly_Salary"}, inplace=True)
print(df)
```

```python
#Q8 Count how many employees belong to each city.
df["City"].value_counts()
```

```python
#Q9 Delete the Age column.
df.drop("Age", axis=1, inplace=True)
print(df)
```

```python
#10 Find employees whose salary is between 30000 and 50000.
df[(df["Salary"] >= 30000) & (df["Salary"] <= 50000)]
```
