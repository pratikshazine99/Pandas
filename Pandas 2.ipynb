```python
import pandas as pd

data = {
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Customer": [" rahul ", "AMAN", "Priya", "shad", "NEHA", "Aman"],
    "Price": ["2500", "Invalid", "3500", "4500", "2000", "3000"],
    "Order_Date": ["2024-01-10", "15/02/2024", "2024-03-20",
                   "2024-04-15", "2024-05-10", "2024-06-01"],
    "City": ["jaipur", "DELHI", "Pune", "JAIPUR", "mumbai", "delhi"],
    "Email": ["rahul@gmail.com", "aman@yahoo.com",
              "priya@gmail.com", "shad@outlook.com",
              "neha@gmail.com", "aman@yahoo.com"]
}

df = pd.DataFrame(data)

print(df)
```

```python
#1 Remove extra spaces from Customer
df["Customer"] = df["Customer"].str.strip()
print(df)
```

```python
#2 Convert Customer names into Title Case
df["Customer"] = df["Customer"].str.title()
print(df)
```

```python
#3 Standardize City names
df["City"] = df["City"].str.strip().str.title()
print(df)
```

```python
#4 Convert Price into numeric format
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
print(df)
```

```python
#5 Handle invalid Price values
df["Price"] = df["Price"].fillna(0)
print(df)
```

```python
#7 Extract Year, Month and Quarter
df["Year"] = df["Order_Date"].dt.year

df["Month"] = df["Order_Date"].dt.month

df["Quarter"] = df["Order_Date"].dt.quarter
print(df)
```

```python
#8 Extract Email domains
df["Domain"] = df["Email"].str.split("@").str[1]
print(df)
```

```python
#9 Find customers using Gmail
df[df["Email"].str.endswith("@gmail.com")]
print(df)
```

```python
#10 Calculate average order price
df["Price"].mean()
print(df)
```

```python
#11 Find orders placed after March 2024
df[df["Order_Date"] > "2024-03-31"]
print(df)
```

```python
#12 Identify duplicate email addresses
df[df["Email"].duplicated(keep=False)]
print(df)
```
