```python
import pandas as pd
import numpy as np

data = {
    "Employee": ["Rahul", "Aman", "Priya", "Shad",
                 "Neha", "Vikas", "Rohit", "Anjali",
                 "Karan", "Simran", "Mohit", "Riya"],

    "Department": ["IT", "HR", "IT", "Finance",
                   "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "HR", "IT"],

    "City": ["Jaipur", "Delhi", "Pune", "Jaipur",
             "Mumbai", "Delhi", "Pune", "Mumbai",
             "Jaipur", "Delhi", "Pune", "Mumbai"],

    "Salary": [25000, 35000, 30000, 45000,
               40000, 55000, 50000, 32000,
               60000, 42000, 38000, 48000],

    "Experience": [1, 3, 2, 5, 4, 7, 6, 2, 8, 5, 3, 6],

    "Gender": ["Male", "Male", "Female", "Male",
               "Female", "Male", "Male", "Female",
               "Male", "Female", "Male", "Female"],

    "Performance": [75, 80, 85, 90, 70, 95,
                    88, 78, 92, 89, 76, 94]
}

df = pd.DataFrame(data)
df
```

### Level 1

```python
#Q1. Calculate total salary department-wise.
df.groupby("Department")["Salary"].sum()
```

```python
#Q2. Find average salary of each department.
df.groupby("Department")["Salary"].mean()
```

```python
#Q3. Find maximum salary in each department.
df.groupby("Department")["Salary"].max()
```

```python
#Q4. Count employees in each department.
df.groupby("Department")["Employee"].count()
```

```python
#Q5. Find average experience city-wise.
df.groupby("City")["Experience"].mean()
```

### Level 2 - Aggregation

```python
#Q6. Calculate sum, mean, median, minimum and maximum salary using agg()
df["Salary"].agg(["sum", "mean", "median", "min", "max"])
```

```python
#Q7. Create a report containing:Total salary,Average salary,Maximum salary,Employee count
report = df.groupby("Department").agg(
    Total_Salary=("Salary", "sum"),
    Average_Salary=("Salary", "mean"),
    Maximum_Salary=("Salary", "max"),
    Employee_Count=("Employee", "count")
)

print(report)
```

```python
#Q8. Group by Department and Gender and calculate average salary.
df.groupby(["Department", "Gender"])["Salary"].mean()
```

```python
#Q9. Find total salary and average performance department-wise.
df.groupby("Department").agg(
    Total_Salary=("Salary", "sum"),
    Average_Performance=("Performance", "mean")
)
```

```python
#Q10. Sort departments by total salary in descending order.
df.groupby("Department")["Salary"].sum().sort_values(ascending=False)
```

### Level 3 - Transform, Filter & Ranking

```python
#Q11. Calculate department-wise average salary for every employee using transform().
df["Dept_Avg_Salary"] = df.groupby("Department")["Salary"].transform("mean")

print(df[["Employee", "Department", "Salary", "Dept_Avg_Salary"]])
```

```python
#Q12. Identify employees earning above their department's average salary.
dept_avg = df.groupby("Department")["Salary"].transform("mean")

result = df[df["Salary"] > dept_avg]

print(result)
```

```python
#Q13. Rank employees by salary within each department.
df["Salary_Rank"] = df.groupby("Department")["Salary"].rank(
    ascending=False
)

print(df[["Employee", "Department", "Salary", "Salary_Rank"]])
```

```python
#Q14. Display only departments having more than 3 employees using filter().
result = df.groupby("Department").filter(
    lambda x: len(x) > 3
)

print(result)
```

```python
#Q15. Find the top 2 highest-paid employees from each department.
result = df.groupby("Department", group_keys=False).apply(
    lambda x: x.nlargest(2, "Salary")
)

print(result)
```

### Level 4 - Business Analyst Challenges

```python
#Q16. Which department has the highest total salary expense?
result = df.groupby("Department")["Salary"].sum()

print(result)
print("Highest Salary Expense:", result.idxmax())
```

```python
#Q17. Which city has the highest average employee salary?
result = df.groupby("City")["Salary"].mean()

print(result)
print("Highest Average Salary:", result.idxmax())
```

```python
#Q18. Find the department with the highest average performance score.
result = df.groupby("Department")["Performance"].mean()

print(result)
print("Highest Average Performance:", result.idxmax())
```

```python
#Q19. Create a pivot table showing average performance by Department and Gender.
result = pd.pivot_table(
    df,
    values="Performance",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)

print(result)
```

```python
#Q20. Create a crosstab showing employee count by City and Department.
result = pd.crosstab(
    df["City"],
    df["Department"]
)

print(result)
```
