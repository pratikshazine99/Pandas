```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../industry_retail_orders_dirty_1000 (1).csv")

df.head()
```

```python
df.isnull().sum()
```

```python
# Null Value Filling
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].mean())
```

```python
print(df.isnull().sum())
```

```python
df["customer_segment"] = df["customer_segment"].fillna(df["customer_segment"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df["product_category"] = df["product_category"].fillna(df["product_category"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df["discount_pct"] = df["discount_pct"].fillna(df["discount_pct"].median())
```

```python
print(df.isnull().sum())
```

```python
df["shipping_fee"] = df["shipping_fee"].fillna(df["shipping_fee"].median())
```

```python
print(df.isnull().sum())
```

```python
df["payment_method"] = df["payment_method"].fillna(df["payment_method"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df["delivery_days"] = df["delivery_days"].fillna(df["delivery_days"].median())
```

```python
print(df.isnull().sum())
```

```python
df["customer_rating"] = df["customer_rating"].fillna(df["customer_rating"].median())
```

```python
print(df.isnull().sum())
```

```python
df["marketing_campaign"] = df["marketing_campaign"].fillna(df["marketing_campaign"].mode()[0])
```

```python
print(df.isnull().sum())
```
